## Example 14: Changing JSON to Dictionary

**Theory and context:** This example belongs to the topic **Changing JSON to Dictionary**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: To change a JSON to a dictionary, first we import the json module and then we use _loads_ method.

````python
import json
# JSON
person_json = '''{
    "name": "veera",
    "country": "Finland",
    "city": "Helsinki",
    "skills": ["JavaScrip", "React", "Python"]
}'''
# let's change JSON to dictionary
person_dct = json.loads(person_json)
print(type(person_dct))
print(person_dct)
print(person_dct['name'])
````

## Example 15: Changing Dictionary to JSON

**Theory and context:** This example belongs to the topic **Changing Dictionary to JSON**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: To change a dictionary to a JSON we use _dumps_ method from the json module.

````python
import json
# python dictionary
person = {
    "name": "veera",
    "country": "Finland",
    "city": "Helsinki",
    "skills": ["JavaScrip", "React", "Python"]
}
# let's convert it to  json
person_json = json.dumps(person, indent=4) # indent could be 2, 4, 8. It beautifies the json
print(type(person_json))
print(person_json)
````

## Example 16: Saving as JSON File

**Theory and context:** This example belongs to the topic **Saving as JSON File**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: We can also save our data as a json file. Let us save it as a json file using the following steps. For writing a json file, we use the json.dump() method, it can take dictionary, output file, ensure_ascii and indent.

````python
import json
# python dictionary
person = {
    "name": "veera",
    "country": "Finland",
    "city": "Helsinki",
    "skills": ["JavaScrip", "React", "Python"]
}
with open('./sample_files/json_example.json', 'w', encoding='utf-8') as f:
    json.dump(person, f, ensure_ascii=False, indent=4)
````

## Example 17: File with csv Extension

**Theory and context:** This example belongs to the topic **File with csv Extension**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: ```csv
"name","country","city","skills"
"veera","Finland","Helsinki","JavaScript"
```
**Example:**

````python
import csv
with open('./sample_files/csv_example.csv') as f:
    csv_reader = csv.reader(f, delimiter=',') # we use, reader method to read csv
    line_count = 0
    for row in csv_reader:
        if line_count == 0:
            print(f'Column names are :{", ".join(row)}')
            line_count += 1
        else:
            print(
                f'\t{row[0]} is a teachers. He lives in {row[1]}, {row[2]}.')
            line_count += 1
    print(f'Number of lines:  {line_count}')
````
