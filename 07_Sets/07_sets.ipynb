## Example 31: Finding Symmetric Difference Between Two Sets

**Theory and context:** This example belongs to the topic **Finding Symmetric Difference Between Two Sets**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: It returns the symmetric difference between two sets. It means that it returns a set that contains all items from both sets, except items that are present in both sets, mathematically: (A\B) âˆª (B\A)

```python
# syntax
st1 = {'item1', 'item2', 'item3', 'item4'}
st2 = {'item2', 'item3'}
# it means (A\B)âˆª(B\A)
st2.symmetric_difference(st1) # {'item1', 'item4'} : st2 ^ st1
```
