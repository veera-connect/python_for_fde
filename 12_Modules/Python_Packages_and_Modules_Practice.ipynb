## 1. What is a module?

A **module** is usually one Python file (`.py`) containing variables, functions, or classes.

Imagine a small shop calculating an order total. We can write the calculation in one Python file and reuse it whenever needed.

```python
# A simple function in the current notebook
def calculate_total(price, quantity):
    return price * quantity

calculate_total(250, 3)
```

Expected result: `750`

If we save reusable functions in a file such as `order_tools.py`, that file can be imported as a module.

## 2. Create and import a custom module

The next cell creates a file named `order_tools.py` in the notebook's current folder. Run it first, then run the import cell below.

**Important:** If you edit the module file after importing it, you may need to restart the kernel or use `importlib.reload()`.

```python
%%writefile order_tools.py
def calculate_total(price, quantity):
    return price * quantity

def apply_discount(amount, discount_percent):
    return amount * (1 - discount_percent / 100)
```

```python
import order_tools

print(order_tools.calculate_total(250, 3))
print(order_tools.apply_discount(750, 10))
```

**Layman explanation:** `import order_tools` makes the file available. The dot (`.`) lets us access a function inside that module, such as `order_tools.calculate_total(...)`.

If Python says `ModuleNotFoundError`, check that the file is in the current working directory and that its name is spelled correctly.

## 3. Explore a module with `dir()`

`dir(module_name)` lists names available in a module. The list can include functions, classes, and internal attributes.

```python
import random

# Show some names available in the random module
[name for name in dir(random) if not name.startswith("_")][:25]
```

## 4. Use `help()` to learn a function

- `random.randint(a, b)` returns an integer from `a` through `b`, including both endpoints.
- Function calls need parentheses and the required arguments.

```python
import random

print("Random number from 1 to 10:", random.randint(1, 10))
help(random.randint)
```

## 5. More useful functions from `random`

Imagine selecting a prize winner or shuffling a list of tasks.

```python
import random

participants = ["Aarav", "Meera", "Kabir", "Isha", "Rohan"]

print("One selected participant:", random.choice(participants))
print("Two selections (with possible repeats):", random.choices(participants, k=2))

random.shuffle(participants)  # Changes the list in place
print("Shuffled list:", participants)
```

## 6. The `math` module

The `math` module provides mathematical constants and functions. It is included with Python, so you do not need to install it.

```python
import math

print("Pi:", math.pi)
print("Square root of 81:", math.sqrt(81))
print("2 raised to power 5:", math.pow(2, 5))
print("Ceiling of 4.2:", math.ceil(4.2))
print("Value of e:", math.e)
```

## 7. The `os` module

The `os` module helps Python interact with the operating system. The examples below only inspect the environment and do not create or delete folders.

```python
import os

print("Current working directory:", os.getcwd())
print("A few entries in the current directory:", os.listdir(".")[:10])
```

## 8. The `time` module

`time.sleep(seconds)` pauses the program. Keep the pause short during demonstrations.

```python
import time

print("Starting...")
time.sleep(1)
print("Finished after a one-second pause.")
```

## 9. The `keyword` module

Python reserves certain words for the language itself, such as `if`, `else`, `for`, and `class`. These are called **keywords**.

```python
import keyword

print("Is 'for' a Python keyword?", keyword.iskeyword("for"))
print("Is 'customer_name' a Python keyword?", keyword.iskeyword("customer_name"))
print("First 10 keywords:", keyword.kwlist[:10])
```

## 10. What is a package?

A **package** organizes related modules into a directory. For example, an imaginary `store_tools` package might contain:
- `pricing.py` — price and discount functions
- `inventory.py` — stock-related functions
- `shipping.py` — delivery calculations

**Hierarchy:** Package → Modules → functions/classes/variables.

Python's import system supports packages; third-party packages can be installed with tools such as `pip`.

## 11. Built-in modules vs third-party packages

| Type | Examples | Installation |
|---|---|---|
| Standard-library module | `random`, `math`, `time`, `os`, `keyword` | Included with Python |
| Third-party package | `requests`, `flask`, `wordcloud` | Usually installed separately |

A package being installed in one Python environment does not guarantee that another Jupyter kernel can import it.

```python
# Check which Python executable this notebook is using
import sys
print(sys.executable)
print("Python version:", sys.version.split()[0])
```

## 13. Common errors and how to fix them

**`ModuleNotFoundError`**
- Confirm the package/module is installed in the environment used by this notebook.
- Check spelling and capitalization.
- For your own module, confirm the `.py` file is in the current folder or Python's import path.
- If you installed a package while Jupyter was open, restart the kernel if needed.

**`AttributeError`**
- The module may not contain the name you typed. Use `dir(module_name)` to inspect available names.

**`TypeError`**
- A function may require arguments. Use `help(module_name.function_name)` to check how to call it.
