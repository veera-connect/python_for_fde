## Example 37: Modifying column values

**Theory and context:** This example belongs to the topic **Modifying column values**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

````python
# Using functions makes our code clean, but you can calculate the bmi without one
def calculate_bmi ():
    weights = df['Weight']
    heights = df['Height']
    bmi = []
    for w,h in zip(weights, heights):
        b = w/(h*h)
        bmi.append(b)
    return bmi

bmi = calculate_bmi()
````
