## Example 7: Accessing Dictionary Items

**Theory and context:** This example belongs to the topic **Accessing Dictionary Items**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: Accessing an item by key name raises an error if the key does not exist. To avoid this error first we have to check if a key exist or we can use the _get_ method. The get method returns None, which is a NoneType object data type, if the key does not exist.

```python
person = {
    'first_name':'veera',
    'last_name':'narni',
    'age':250,
    'country':'Finland',
    'is_marred':True,
    'skills':['JavaScript', 'React', 'Node', 'MongoDB', 'Python'],
    'address':{
        'street':'Space street',
        'zipcode':'02210'
    }
    }
print(person.get('first_name')) # veera
print(person.get('country'))    # Finland
print(person.get('skills')) #['JavaScript', 'React', 'Node', 'MongoDB', 'Python']
print(person.get('city'))   # None
```

## Example 13: Removing Key and Value Pairs from a Dictionary

**Theory and context:** This example belongs to the topic **Removing Key and Value Pairs from a Dictionary**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: - _pop(key)_: removes the item with the specified key name:
- _popitem()_: removes the last item
- _del_: removes an item with specified key name

```python
# syntax
dct = {'key1':'value1', 'key2':'value2', 'key3':'value3', 'key4':'value4'}
dct.pop('key1') # removes key1 item
dct = {'key1':'value1', 'key2':'value2', 'key3':'value3', 'key4':'value4'}
dct.popitem() # removes the last item
del dct['key2'] # removes key2 item
```
