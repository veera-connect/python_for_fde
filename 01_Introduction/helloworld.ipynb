## 3. Arithmetic operators

**Theory:** Arithmetic operators calculate with numbers. Use them for totals, measurements, prices, and formulas; compare `/`, `//`, `%`, and `**`.

**When to use it:** Use this technique in a program whenever you need the behavior shown in the example. Try changing the inputs to see how the result changes.

```python
# Arithmetic operators
print(2 + 3)   # addition(+)
print(3 - 1)   # subtraction(-)
print(2 * 3)   # multiplication(*)
print(3 / 2)   # division(/)
print(3 ** 2)  # exponential(**)
print(3 % 2)   # modulus(%)
print(3 // 2)  # floor division(//)
```

## 4. Checking data types

**Theory:** A type tells Python what kind of value it is handling, such as text, a number, or a list. Use `type()` to inspect values while learning or debugging.

**When to use it:** Use this technique in a program whenever you need the behavior shown in the example. Try changing the inputs to see how the result changes.

```python
# Checking data types
print(type(10))                  # int
print(type(3.14))                # float
print(type(1 + 3j))              # complex
print(type('veera'))             # str
print(type(True))                # bool
print(type([1, 2, 3]))           # list
print(type((9.8, 3.14, 2.7)))    # tuple
print(type({9.8, 3.14, 2.7}))    # set
print(type({'name': 'veera'}))   # dict
print(type(None))                # NoneType
```
