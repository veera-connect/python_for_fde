## Example 1: Exception Handling

**Theory and context:** This example belongs to the topic **Exception Handling**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: Python uses _try_ and _except_ to handle errors gracefully. A graceful exit (or graceful handling) of errors is a simple programming idiom - a program detects a serious error condition and "exits gracefully", in a controlled manner as a result. Often the program prints a descriptive error message to a terminal or log as part of the graceful exit, this makes our application more robust. The cause of an exception is often external to the program itself. An example of exceptions could be an incorrect input, wrong file name, unable to find a file, a malfunctioning IO device. Graceful handling of errors prevents our applications from crashing.
We have covered the different Python _error_ types in the previous section. If we use _try_ and _except_ in our program, then it will not raise errors in those blocks.
The full skeleton uses _try_, _except_, _else_ and _finally_:

````python
try:
    # code that might raise an error
except SomeError:
    # runs only if a matching error was raised
else:
    # runs only if the try block did NOT raise an error
finally:
    # always runs, whether or not an error was raised
````

## Example 2: Exception Handling

**Theory and context:** This example belongs to the topic **Exception Handling**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: **Example:** the flow of `try` / `except` / `else` / `finally`.

````python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError as e:
        print('except: cannot divide by zero ->', e)
    else:
        print('else: division succeeded, result =', result)
    finally:
        print('finally: this always runs')

print('--- veera divides 10 by 2 ---')
divide(10, 2)
print('--- veera divides 10 by 0 ---')
divide(10, 0)
````

## Example 10: Unpacking Lists

From the lesson: When we run the this code, it raises an error, because this function takes numbers (not a list) as arguments. Let us unpack/destructure the list.

````python
def sum_of_five_nums(a, b, c, d, e):
    return a + b + c + d + e

lst = [1, 2, 3, 4, 5]
print(sum_of_five_nums(*lst))  # 15
````

## Example 11: Unpacking Lists

**Theory and context:** This example belongs to the topic **Unpacking Lists**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: We can also use unpacking in the range built-in function that expects a start and an end.

````python
numbers = range(2, 7)  # normal call with separate arguments
print(list(numbers)) # [2, 3, 4, 5, 6]
args = [2, 7]
numbers = range(*args)  # call with arguments unpacked from a list
print(numbers)      # [2, 3, 4, 5,6]
````

## Example 13: Unpacking Dictionaries

**Theory and context:** This example belongs to the topic **Unpacking Dictionaries**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

````python
def unpacking_person_info(name, country, city, age):
    return f'{name} lives in {country}, {city}. He is {age} year old.'
dct = {'name':'veera', 'country':'Finland', 'city':'Helsinki', 'age':250}
print(unpacking_person_info(**dct)) # veera lives in Finland, Helsinki. He is 250 years old.
````

## Example 14: Packing Lists

**Theory and context:** This example belongs to the topic **Packing Lists**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

````python
def sum_all(*args):
    s = 0
    for i in args:
        s += i
    return s
print(sum_all(1, 2, 3))             # 6
print(sum_all(1, 2, 3, 4, 5, 6, 7)) # 28
````

````python
def packing_person_info(**kwargs):
    # check the type of kwargs and it is a dict type
    # print(type(kwargs))
    # Printing dictionary items
    for key in kwargs:
        print(f"{key} = {kwargs[key]}")
    return kwargs

print(packing_person_info(name="veera",
      country="Finland", city="Helsinki", age=250))
````

## Example 16: Spreading in Python

**Theory and context:** This example belongs to the topic **Spreading in Python**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: Like in JavaScript, spreading is possible in Python. Let us check it in an example below:

````python
lst_one = [1, 2, 3]
lst_two = [4, 5, 6, 7]
lst = [0, *lst_one, *lst_two]
print(lst)          # [0, 1, 2, 3, 4, 5, 6, 7]
country_lst_one = ['Finland', 'Sweden', 'Norway']
country_lst_two = ['Denmark', 'Iceland']
nordic_countries = [*country_lst_one, *country_lst_two]
print(nordic_countries)  # ['Finland', 'Sweden', 'Norway', 'Denmark', 'Iceland']
````

## Example 19: Zip

**Theory and context:** This example belongs to the topic **Zip**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: Sometimes we would like to combine lists when looping through them. See the example below:

````python
fruits = ['banana', 'orange', 'mango', 'lemon', 'lime']                    
vegetables = ['Tomato', 'Potato', 'Cabbage','Onion', 'Carrot']
fruits_and_veges = []
for f, v in zip(fruits, vegetables):
    fruits_and_veges.append({'fruit':f, 'veg':v})

print(fruits_and_veges)
````
