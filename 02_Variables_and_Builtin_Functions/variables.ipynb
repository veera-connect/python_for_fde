Variables store values under readable names so a program can reuse information. Built-in functions help inspect or display those values.

## 1. Variables in Python

**Theory:** Variables are names for values your program needs to remember. Use clear names to make code easier to read and reuse.

**When to use it:** Use this technique in a program whenever you need the behavior shown in the example. Try changing the inputs to see how the result changes.

```python
# Variables in Python

first_name = 'veera'
last_name = 'narni'
country = 'Finland'
city = 'Helsinki'
age = 250
is_married = True
skills = ['HTML', 'CSS', 'JS', 'React', 'Python']
person_info = {
    'firstname': 'veera',
    'lastname': 'narni',
    'country': 'Finland',
    'city': 'Helsinki'
}
```

## 2. Printing the values stored in the variables

**Theory:** `print()` displays information in the output area. Use it for messages, results, and quick checks while learning.

**When to use it:** Use this technique in a program whenever you need the behavior shown in the example. Try changing the inputs to see how the result changes.

```python
# Printing the values stored in the variables

print('First name:', first_name)
print('First name length:', len(first_name))
print('Last name: ', last_name)
print('Last name length: ', len(last_name))
print('Country: ', country)
print('City: ', city)
print('Age: ', age)
print('Married: ', is_married)
print('Skills: ', skills)
print('Person information: ', person_info)
```

## 3. Declaring multiple variables in one line

**Theory:** Variables are names for values your program needs to remember. Use clear names to make code easier to read and reuse.

**When to use it:** Use this technique in a program whenever you need the behavior shown in the example. Try changing the inputs to see how the result changes.

```python
# Declaring multiple variables in one line

first_name, last_name, country, age, is_married = 'veera', 'narni', 'Helsink', 250, True

print(first_name, last_name, country, age, is_married)
print('First name:', first_name)
print('Last name: ', last_name)
print('Country: ', country)
print('Age: ', age)
print('Married: ', is_married)
```
