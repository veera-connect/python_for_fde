## Example 2: Getting *datetime* Information

**Theory and context:** This example belongs to the topic **Getting *datetime* Information**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

```python
from datetime import datetime
now = datetime.now()
print(now)                      # 2021-07-08 07:34:46.549883
day = now.day                   # 8
month = now.month               # 7
year = now.year                 # 2021
hour = now.hour                 # 7
minute = now.minute             # 38
second = now.second
timestamp = now.timestamp()
print(day, month, year, hour, minute)
print('timestamp', timestamp)
print(f'{day}/{month}/{year}, {hour}:{minute}')  # 8/7/2021, 7:38
```

## Example 4: Formatting Date Output Using *strftime*

**Theory and context:** This example belongs to the topic **Formatting Date Output Using *strftime***. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: Formatting date time using *strftime* method and the documentation can be found [here](https://strftime.org/).

```python
from datetime import datetime
# current date and time
now = datetime.now()
t = now.strftime("%H:%M:%S")
print("time:", t)           # time: 18:21:40
time_one = now.strftime("%m/%d/%Y, %H:%M:%S")
# mm/dd/YY H:M:S format
print("time one:", time_one)        # time one: 06/28/2022, 18:21:40
time_two = now.strftime("%d/%m/%Y, %H:%M:%S")
# dd/mm/YY H:M:S format
print("time two:", time_two)        # time two: 28/06/2022, 18:21:40
```

## Example 5: Formatting Date Output Using *strftime*

**Theory and context:** This example belongs to the topic **Formatting Date Output Using *strftime***. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: | `%x` | Locale's date representation | 09/16/26 |
| `%X` | Locale's time representation | 21:55:49 |
| `%Z` | Time zone name | (empty if naive) |
| `%z` | UTC offset | (empty if naive) |
**Example:**

```python
from datetime import datetime

now = datetime.now()
print(now.strftime("%a"))    # Wed
print(now.strftime("%A"))    # Wednesday
print(now.strftime("%B"))    # September
print(now.strftime("%Y"))    # 2026
print(now.strftime("%p"))    # PM

formatted = now.strftime("%A, %d %B %Y %I:%M %p")
print("formatted:", formatted)
```

## Example 6: String to Time Using *strptime*

**Theory and context:** This example belongs to the topic **String to Time Using *strptime***. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

From the lesson: Here is a [documentation](https://www.programiz.com/python-programming/datetime/strptime) hat helps to understand the format.

```python
from datetime import datetime
date_string = "5 December, 2019"
print("date_string =", date_string)     # date_string = 5 December, 2019
date_object = datetime.strptime(date_string, "%d %B, %Y")
print("date_object =", date_object)     # date_object = 2019-12-05 00:00:00
```

## Example 8: Time Objects to Represent Time

**Theory and context:** This example belongs to the topic **Time Objects to Represent Time**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

```python
from datetime import time
# time(hour = 0, minute = 0, second = 0)
a = time()
print("a =", a)     # a = 00:00:00
# time(hour, minute and second)
b = time(10, 30, 50)
print("b =", b)     # b = 10:30:50
# time(hour, minute and second)
c = time(hour=10, minute=30, second=50)
print("c =", c)     # c = 10:30:50
# time(hour, minute, second, microsecond)
d = time(10, 30, 50, 200555)
print("d =", d)     # d = 10:30:50.200555
```

## Example 9: Difference Between Two Points in Time Using

**Theory and context:** This example belongs to the topic **Difference Between Two Points in Time Using**. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

```python
from datetime import date, datetime
today = date(year=2019, month=12, day=5)
new_year = date(year=2020, month=1, day=1)
time_left_for_newyear = new_year - today
# Time left for new year:  27 days, 0:00:00
print('Time left for new year: ', time_left_for_newyear)  # Time left for new year:  27 days, 0:00:00

t1 = datetime(year = 2019, month = 12, day = 5, hour = 0, minute = 59, second = 0)
t2 = datetime(year = 2020, month = 1, day = 1, hour = 0, minute = 0, second = 0)
diff = t2 - t1
print('Time left for new year:', diff) # Time left for new year: 26 days, 23: 01: 00
```

## Example 10: Difference Between Two Points in Time Using *timedelta*

**Theory and context:** This example belongs to the topic **Difference Between Two Points in Time Using *timedelta***. Use the technique when your program needs to perform this kind of task. Try changing an input and compare the result.

```python
from datetime import timedelta
t1 = timedelta(weeks=12, days=10, hours=4, seconds=20)
t2 = timedelta(days=7, hours=5, minutes=3, seconds=30)
t3 = t1 - t2
print("t3 =", t3)
```
